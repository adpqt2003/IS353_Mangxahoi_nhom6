```python
mamh_mapping = dict(zip(mamh_df['org_id'], mamh_df['remap_id']))
mssv_mapping = dict(zip(mssv_df['org_id'], mssv_df['remap_id']))
```

```python
import polars as pl

interactions = pl.DataFrame(interactions)
```

```python
# Mapping và group theo mssv
interactions = interactions.with_columns(
    pl.col('mssv').replace(mssv_mapping).cast(pl.Int32),
    pl.col('mamh').replace(mamh_mapping).cast(pl.Int32))\
    .group_by('mssv') \
    .agg(pl.col('mamh')) \
    .rename({'mamh': 'monhoc_order'}) \
    .sort('mssv')
```

```python
interactions
```

```python
import tqdm

# Mở file để ghi dữ liệu
train_file = open('/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Train_test_data/MODEL_DATA/train.txt', 'w')
test_file = open('/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Train_test_data/MODEL_DATA/test.txt', 'w')

# Duyệt qua từng dòng dữ liệu trong sorted_interactions
for row in tqdm.tqdm(interactions.iter_rows(named=True)):
    user = row['mssv']
    courses = row['monhoc_order']

    # Kiểm tra nếu người dùng có hơn một khóa học
    if len(courses) > 1:
        # Chia dữ liệu thành train, val, test
        train_courses = ' '.join(map(str, courses[:-1]))
        test_course = courses[-1]

        train_file.write(f"{user} {train_courses}\n")
        test_file.write(f"{user} {test_course}\n")
    else:
        # Nếu chỉ có một khóa học, ghi vào test
        train_file.write(f"{user} {courses[0]}\n")
        test_file.write(f"{user} {courses[0]}\n")

# Đóng các file sau khi ghi
train_file.close()
test_file.close()
```
