# Xử lý dữ liệu

### Cột hocky

```python
# kiểm tra dữ liệu khuyết
missing_count = df['hocky'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'hocky': {missing_count}")
```

### cột namhoc

```python
# Số lượng dữ liệu khuyết
missing_count = df['namhoc'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'namhoc': {missing_count}")
```

### cột drl

```python
# Số lượng khuyết
missing_count = df['drl'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'drl': {missing_count}")
```

### cột ghichu

```python
# Số lượng khuyết
missing_count = df['ghichu'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'ghichu': {missing_count}")
```

```python
# Hàm gán giá trị vào cột ghichu dựa trên giá trị drl
def assign_rating(drl):
  '''
    Hàm gán giá trị vào cột ghichu dựa trên giá trị drl
    Param:
      drl: Giá trị có trong cột drl
  '''
  if drl >= 90:
      return 'Xuất sắc'
  elif 80 < drl <= 90:
      return 'Tốt'
  elif 65 < drl <= 80:
      return 'Khá'
  elif 50 < drl <= 65:
      return 'Trung bình'
  else:
      return 'Yếu'

# Cập nhật cột ghichu
df['ghichu'] = df['drl'].apply(assign_rating)
```

```python
# Kiểm tra giá trị duy nhất có trong cột ghi chú
df['ghichu'].unique()
```

```python
# Bỏ các ký tự đặc biệt, chữ thường, xóa khoảng trắng
excute_column = ['mssv']
for column  in df.select_dtypes(include=['object']).columns:
  if column not in excute_column:
    df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%]", '', regex=True)
```

```python
# Xóa khoảng trắng thừa trong cột "ghichu"
df['ghichu'] = df['ghichu'].str.strip()  # Xóa khoảng trắng đầu và cuối
df['ghichu'] = df['ghichu'].str.replace(r'\s+', '', regex=True)  # Xóa khoảng trắng thừa giữa các từ
```
