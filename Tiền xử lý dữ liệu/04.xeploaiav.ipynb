# Đọc & xử lý dữ liệu

```python
data_path  = "/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Data_clean/EDA"
```

```python
# Load dữ liệu
df = pd.read_excel(data_path + "/04.xeploaiav_EDA.xlsx") # Có thể thay đổi theo shortcut trên drive
```

```python
# Bỏ các ký tự đặc biệt, chữ thường, xóa khoảng trắng
for column in df.select_dtypes(include=['object']).columns:
  if column != 'mssv':
    df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%]", '', regex=True)
```

# Kiểm tra dữ liệu bị khuyết

```python
# Đếm số lượng giá trị bị khuyết cho từng cột
missing_values = df.isnull().sum()
missing_values
```

```python
# Tỷ lệ giá trị thiếu
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)
print("Tỷ lệ giá trị thiếu (%):")
print(missing_percentage)
```

```python
# Kiểm tra số dòng chứa giá trị null
num_null_rows = df.isnull().any(axis=1).sum()
print(f"Số dòng chứa giá trị null: {num_null_rows}")
```

```python
# Lọc các dòng có chứa giá trị null
rows_with_null = df[df.isnull().any(axis=1)]
print(rows_with_null)
```

# Xử lý dữ liệu

### cột mssv

Drop dữ liệu bị khuyết trong total và mssv do dữ liệu bị khuyết rất nhỏ

```python
# Kiểm tra dữ liệu khuyết
missing_count = df['mssv'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mssv': {missing_count}")
```

```python
# Kiểm tra trùng lặp dữ liệu với khóa chính là tổ hợp mssv
duplicate_count = df.duplicated(subset=['mssv']).sum()
print(f"Số giá trị trùng lặp trong tổ hợp 'mssv': {duplicate_count + 1}")
```

```python
# Danh sách các giá trị 'mssv' bị trùng lặp
duplicate_values = df[df.duplicated(subset='mssv', keep=False)]['mssv'].unique()
print("Các giá trị bị trùng lặp:", duplicate_values)
```

```python
df.dropna(subset=['mssv'], inplace=True)
```

### cột mamh

Đối với cột mamh sẽ được điền theo môn xuất hiện nhiều nhất

```python
# Kiểm tra dữ liệu khuyết
missing_count = df['mamh'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mamh': {missing_count}")
```

```python
# Các giá trị duy nhất và tần suất cho các cột định mamh
print("\nGiá trị duy nhất trong cột 'mamh':")
print(df['mamh'].value_counts())
print(df['mamh'].unique())
```

```python
df['mamh'].mode()
```

```python
# Điền dữ liệu khuyết theo mode
df['mamh'].fillna(df['mamh'].mode()[0], inplace=True)
```

### cột ghichu

Do là dữ liệu cột ghi chú chỉ có thông tin về TOEIC nên sẽ chuyển thông tin từ dữ liệu cột ghi chú qua cột mamh vì cột điểm ở cột total tại những dòng có ghi chú là TOEIC thì điểm đó total không còn bằng tổng 'listening', 'reading'

```python
# kiểm tra dữ liệu khuyết
missing_count = df['ghichu'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'ghichu': {missing_count}")
```

```python
df['ghichu'].unique()
```

```python
# Lọc dữ liệu theo ghi chú liên quan đến TOEIC
toeic_conditions = [' qui đổi toeic']
toeic_columns = df[df['ghichu'].isin(toeic_conditions)]
```

```python
# Chuyển giá trị từ cột `ghichu` sang `mamh`
df.loc[df['ghichu'].isin(toeic_conditions), 'mamh'] = 'toeic'
```

```python
df['mamh'].unique()
```

```python
# Loại bỏ cột 'ghichu' khỏi dữ liệu
df.drop(columns=['ghichu'], inplace=True)
```

```python
# Các cột còn lại
df.dtypes
```

### Kiểm tra cột total

```python
# kiểm tra dữ liệu khuyết
missing_count = df['total'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'total': {missing_count}")
```

```python
# Các giá trị duy nhất và tần suất trong cột total
print(f"\nGiá trị duy nhất trong cột 'total':")
print(df['total'].value_counts())
print(df['total'].unique())
```
