**Kiểm tra thuộc tính nơi sinh**

Danh sách nơi sinh có trong tập data

```python
def unique_values_in_column(df, column_name):
    """
    Lấy giá trị duy nhất trong 1 cột giá trị.

    Parameters:
    - df (DataFrame): DataFrame.
    - column_name (str): Tên cột muốn trích xuất dữ liệu.

    Returns:
    - Danh sách các giá trị unique.
    """
    if column_name in df.columns:
        unique_values = df[column_name].dropna().unique()  # Drop NaN và lấy giá trị unique
        return list(unique_values)
    else:
        print(f"Cột '{column_name}' không có trong DataFrame.")
        return None

# Lấy các giá trị unique
column_to_check = 'noisinh'
unique_values = unique_values_in_column(df_filtered, column_to_check)

# In ra màn hình
print(f"Danh sách các giá trị trong cột '{column_to_check}': {unique_values}")
```

```python
if 'noisinh' in df_filtered.columns:
    '''
    Kiểm tra nếu cột noisinh có trong df
    '''
    # Tính tỉ lệ mỗi nơi sinh của sinh viên
    frequency = df_filtered['noisinh'].value_counts()

    # Chuyển kết quả sang dataframe
    frequency_df = frequency.reset_index()
    frequency_df.columns = ['noisinh', 'frequency']

    # Tính tỉ lệ phần trăm
    total = frequency_df['frequency'].sum()
    frequency_df['percentage'] = (frequency_df['frequency'] / total * 100).round(2)

    # In kết quả
    print(frequency_df)
```

Đưa các dữ liệu bị lệch cột ở các thuộc tính hocky, namhoc, soqd, ngayqd, Column1 về đúng vị trí

```python
# Function phát hiện các dữ liệu bị lệch
def fix_misaligned_rows(df):
    # Xác định những dòng có giá trị cột Column1 != NaN
    misaligned_rows = df[df['Column1'].notna()]

    for idx in misaligned_rows.index:
        # Dịch 1 cột trang trái
        df.loc[idx, 'hocky':'Column1'] = df.loc[idx, 'hocky':'Column1'].shift(-1)
        # Chuyển giá trị cột Column1 về NaN
        df.loc[idx, 'Column1'] = np.nan

    return df

# Apply the function to fix the dataset
df = fix_misaligned_rows(df)
df
```

```python
from dateutil.parser import parse

def detect_datetime_formats(df, column_name):
    """
    Xác định các format ngày giờ cho cột dữ liệu nhất định.

    Parameters:
    - df (DataFrame): DataFrame.
    - column_name (str): Tên cột.

    Returns:
    - Danh sách các định dạng ngày giờ.
    """
    if column_name not in df.columns:
        print(f"Cột '{column_name}' không có trong Dataframe.")
        return None

    formats = set()
    for value in df[column_name].dropna():
        try:
            parsed_date = parse(str(value), fuzzy=True)
            # Extract the inferred format
            formats.add(parsed_date.strftime("%Y-%m-%d"))
        except ValueError:
            continue  # Skip non-datetime values

    return list(formats)

column_to_check = 'ngayqd'
datetime_formats = detect_datetime_formats(df_filtered, column_to_check)

# Print detected datetime formats
print(f"Detected datetime formats in column '{column_to_check}': {datetime_formats}")
```

```python
def unique_values_in_column(df, column_name):
    """
    Lấy giá trị duy nhất trong 1 cột giá trị.

    Parameters:
    - df (DataFrame): DataFrame.
    - column_name (str): Tên cột muốn trích xuất dữ liệu.

    Returns:
    - Danh sách các giá trị unique.
    """
    if column_name in df.columns:
        unique_values = df[column_name].dropna().unique()  # Drop NaN và lấy giá trị unique
        return list(unique_values)
    else:
        print(f"Cột '{column_name}' không có trong DataFrame.")
        return None

# Lấy các giá trị unique
column_to_check = 'hocky'
unique_values = unique_values_in_column(df_filtered, column_to_check)

# In ra màn hình
print(f"Danh sách các giá trị trong cột '{column_to_check}': {unique_values}")
```

```python
if 'hocky' in df_filtered.columns:
    '''
    Kiểm tra nếu cột hocky có trong df
    '''
    # Tính tần suất mỗi học kỳ
    frequency = df_filtered['hocky'].value_counts()

    # Chuyển kết quả sang dataframe
    frequency_df = frequency.reset_index()
    frequency_df.columns = ['hocky', 'frequency']

    # Tính tỉ lệ phần trăm
    total = frequency_df['frequency'].sum()
    frequency_df['percentage'] = (frequency_df['frequency'] / total * 100).round(2)

    # In kết quả
    print(frequency_df)
```
