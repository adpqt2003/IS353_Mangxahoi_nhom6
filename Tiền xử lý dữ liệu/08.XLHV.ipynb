# Đọc & xử lý dữ liệu

```python
data_path  = "/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Data_clean/EDA"
```

```python
# Load dữ liệu
df = pd.read_excel(data_path + "/08.XLHV_EDA.xlsx") # Có thể thay đổi theo shortcut trên drive
```

```python
# Loại trừ các cột 'A', 'C', 'D'
exclude_columns = ['mssv', 'lydo', 'ngayqd']

# Chỉ xử lý các cột kiểu object và không nằm trong exclude_columns
for column in df.select_dtypes(include=['object']).columns:
    if column not in exclude_columns:
        df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%\-]", '', regex=True)
```

```python
# Lọc các dòng có chứa giá trị null
rows_with_null = df[df.isnull().any(axis=1)]
print(rows_with_null)
```

# Xử lý dữ liệu

### Cột mssv

```python
# kiểm tra dữ liệu khuyết
missing_count = df['mssv'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mssv': {missing_count}")
```

```python
# Loại bỏ các cột bị khuyết
df.dropna(subset=['mssv'], inplace=True)
```

### cột tinhtrang

```python
# Số lượng dữ liệu khuyết
missing_count = df['tinhtrang'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'tinhtrang': {missing_count}")
```

```python
# Chuyển đổi cột 'namsinh' sang kiểu số nguyên để loại bỏ phần .0
df['tinhtrang'] = df['tinhtrang'].astype('Int64')
```

### cột lydo

```python
# Số lượng khuyết
missing_count = df['lydo'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'lydo': {missing_count}")
```

```python
# Xóa khoảng trắng thừa trong cột "lydo"
df['lydo'] = df['lydo'].str.strip()  # Xóa khoảng trắng đầu và cuối
df['lydo'] = df['lydo'].str.replace(r'\s+', '', regex=True)  # Xóa khoảng trắng thừa giữa các từ
```

```python
# Loại trừ các cột 'A', 'D'
exclude_columns = ['mssv', 'ngayqd']

# Chỉ xử lý các cột kiểu object và không nằm trong exclude_columns
for column in df.select_dtypes(include=['object']).columns:
    if column not in exclude_columns:
        df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%\-]", '', regex=True)
```

## cột hocky

```python
# Số lượng khuyết
missing_count = df['hocky'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'hocky': {missing_count}")
```

```python
# Chuyển đổi cột 'hocky' sang kiểu số nguyên để loại bỏ phần .0
df['hocky'] = df['hocky'].astype('Int64')
```

## cột namhoc

```python
# Số lượng khuyết
missing_count = df['namhoc'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'namhoc': {missing_count}")
```

```python
# Chuyển đổi cột 'namhoc' sang kiểu số nguyên để loại bỏ phần .0
df['namhoc'] = df['namhoc'].astype('Int64')
```

## cột ngayqd

```python
# Kiểm tra giá trị khuyết
missing_count = df['ngayqd'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'ngayqd': {missing_count}")
```

```python
# Xử lý giá trị khuyết bằng các điền mode
df['ngayqd'].fillna(df['ngayqd'].mode()[0], inplace=True)
```

```python
# Chạy lần 2
df['ngayqd'].fillna(df['ngayqd'].mode()[0], inplace=True)
```

```python
# Thay đổi dữ liệu về định dạng chung dd/mm/yyyy
df['ngayqd'] = pd.to_datetime(df['ngayqd'], errors='coerce').dt.strftime('%d/%m/%Y')
```
