# Đọc và xử lý dữ liệu

```python
data_path  = "/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Data_clean/EDA" # Đường dẫn đến Folder Drive lưu file dữ liệu x_EDA.xlsx
```

```python
# Load dữ liệu
df = pd.read_excel(data_path + "/sinhvien_dtb_toankhoa_EDA.xlsx") # Có thể thay đổi theo shortcut trên drive
```

```python
# Bỏ các ký tự đặc biệt, chữ thường, xóa khoảng trắng
for column in df.select_dtypes(include=['object']).columns:
  if column != 'mssv':
    df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%]", '', regex=True)
```

# Kiểm tra dữ liệu bị khuyết

```python
# Đếm số lượng giá trị bị khuyết cho từng cột
missing_values = df.isnull().sum()
missing_values
```

```python
# Tỷ lệ giá trị thiếu
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)
print("Tỷ lệ giá trị thiếu (%):")
print(missing_percentage)
```

```python
# Lọc các dòng có chứa giá trị null
rows_with_null = df[df.isnull().any(axis=1)]
print(rows_with_null)
```

# Xử lý dữ liệu

### Cột dtb_toankhoa

```python
# Xử lý giá trị khuyết bằng cách điền mean
df.fillna({'dtb_toankhoa': df['dtb_toankhoa'].mean()}, inplace=True)
```

### Cột dtb_tichluy

```python
# Xử lý giá trị khuyết bằng cách điền mean
df.fillna({'dtb_tichluy': df['dtb_tichluy'].mean()}, inplace=True)
```
