# Xử lý dữ liệu

### Cột mssv

```python
# kiểm tra dữ liệu khuyết
missing_count = df['mssv'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mssv': {missing_count}")
```

### Cột dien_tt

```python
# kiểm tra dữ liệu khuyết
missing_count = df['dien_tt'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'dien_tt': {missing_count}")
```

###Cột diem_tt

```python
# kiểm tra dữ liệu khuyết
missing_count = df['diem_tt'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'diem_tt': {missing_count}")
```

```python
#Thay thế giá trị khuyết
df['diem_tt'].fillna(df['diem_tt'].mean(), inplace=True)
```

###Cột TEN_TRUONG

```python
# Số lượng khuyết
missing_count = df['TEN_TRUONG'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'TEN_TRUONG': {missing_count}")
```

```python
# Bỏ khoảng trắng trong các tên trường
df['TEN_TRUONG'] = df['TEN_TRUONG'].str.replace(" ", "")

# Các giá trị duy nhất và tần suất trong cột TEN_TRUONG
print(f"\nGiá trị duy nhất trong cột 'TEN_TRUONG':")
print(df['TEN_TRUONG'].value_counts())
print(df['TEN_TRUONG'].unique())
```

```python
# Chuyển dữ liệu thành chữ thường và xóa khoản trắng
df['TEN_TRUONG'] = df['TEN_TRUONG'].str.strip().str.lower()
```

```python
# Thay thế giá trị rỗng hoặc chỉ chứa khoảng trắng thành 'Không điền'
df['TEN_TRUONG'] = df['TEN_TRUONG'].replace(r'^\s*$', 'không', regex=True)
df = df.fillna('Không điền')
```
