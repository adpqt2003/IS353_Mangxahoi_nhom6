# Đọc & xử lý dữ liệu

```python
data_path  = "/content/drive/MyDrive/IS353 - Mạng xã hội - Nhóm 6/Đồ án môn học/Project/Data/Data_clean/EDA"
```

```python
# Load dữ liệu
df = pd.read_excel(data_path + "/01.sinhvien_EDA.xlsx") # Có thể thay đổi theo shortcut trên drive
```

```python
# Bỏ các ký tự đặc biệt, chữ thường, xóa khoảng trắng
for column in df.select_dtypes(include=['object']).columns:
  if column != 'mssv':
    df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%]", '', regex=True)
```

# Kiểm tra dữ liệu bị khuyết

```python
# Đếm số lượng giá trị bị khuyết cho từng cột
missing_values = df.isnull().sum()
missing_values
```

```python
# Tỷ lệ giá trị thiếu
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)
print("Tỷ lệ giá trị thiếu (%):")
print(missing_percentage)
```

```python
# Kiểm tra số dòng chứa giá trị null
num_null_rows = df.isnull().any(axis=1).sum()
print(f"Số dòng chứa giá trị null: {num_null_rows}")
```

```python
# Lọc các dòng có chứa giá trị null
rows_with_null = df[df.isnull().any(axis=1)]
print(rows_with_null)
```

# Xử lý dữ liệu

### Cột mssv

```python
# Kiểm tra trùng lặp dữ liệu với khóa chính là tổ hợp mssv
duplicate_count = df.duplicated(subset=['mssv']).sum()
print(f"Số giá trị trùng lặp trong tổ hợp 'mssv': {duplicate_count + 1}")
```

```python
# Danh sách các giá trị 'mssv' bị trùng lặp
duplicate_values = df[df.duplicated(subset='mssv', keep=False)]['mssv'].unique()
print("Các giá trị bị trùng lặp:", duplicate_values)
```

```python
# kiểm tra dữ liệu khuyết
missing_count = df['mssv'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mssv': {missing_count}")
```

```python
# Loại bỏ các cột bị khuyết
df.dropna(subset=['mssv'], inplace=True)
```

### cột namsinh

```python
# Số lượng dữ liệu khuyết
missing_count = df['namsinh'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'namsinh': {missing_count}")
```

```python
# Xử lý giá trị khuyết bằng các điền mode
df['namsinh'].fillna(df['namsinh'].mode()[0], inplace=True)
```

```python
# Chuyển đổi cột 'namsinh' sang kiểu số nguyên để loại bỏ phần .0
df['namsinh'] = df['namsinh'].astype('Int64')
```

```python
# Chuyển đổi cột 'gioitinh' sang kiểu số nguyên để loại bỏ phần .0
df['gioitinh'] = df['gioitinh'].astype('Int64')
```

### cột noisinh

```python
# Kiểm tra giá trị khuyết
missing_count = df['noisinh'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'noisinh': {missing_count}")
```

```python
# Bỏ khoảng trắng trong các nơi sinh
df['noisinh'] = df['noisinh'].str.replace(" ", "")

# Các giá trị duy nhất và tần suất trong cột noisinh
print(f"\nGiá trị duy nhất trong cột 'noisinh':")
print(df['noisinh'].value_counts())
print(df['noisinh'].unique())
```

```python
# Chuyển dữ liệu thành chữ thường và xóa khoản trắng
df['noisinh'] = df['noisinh'].str.strip().str.lower()
```

```python
# Thay đổi dữ liệu về định dạng chung
df['noisinh'] = df['noisinh'].str.replace("'", "")\
                                  .str.replace("thành phố ", "")\
                                  .str.replace("tp", "")\
                                  .str.replace(".", "")\
                                  .str.replace("thành phố hồ chí minh", "hồ chí minh")\
                                  .str.replace("thànhphốhồchíminh", "hồ chí minh")\
                                  .str.replace("city", "")\
                                  .str.replace("null", "")\
                                  .str.strip()
```

```python
# Thay thế giá trị rỗng hoặc chỉ chứa khoảng trắng thành 'không'
df['noisinh'] = df['noisinh'].replace(r'^\s*$', 'không', regex=True)
```
