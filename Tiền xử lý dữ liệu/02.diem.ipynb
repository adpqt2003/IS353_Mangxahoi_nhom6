# Đọc & xử lý dữ liệu

```python
data_path = '/content/drive/MyDrive/Data/Data_clean/EDA'
```

```python
# Load dữ liệu
df = pd.read_excel(data_path + '/02.diem_EDA.xlsx')
```

```python
# Bỏ các ký tự đặc biệt, chữ thường, xóa khoảng trắng
excute_column = ['mssv']
for column  in df.select_dtypes(include=['object']).columns:
  if column not in excute_column:
    df[column] = df[column].str.lower().str.replace(r"[',:);*@$/%]", '', regex=True)
```

# Kiểm tra dữ liệu bị khuyết

```python
# Đếm số lượng giá trị bị khuyết cho từng cột
missing_values = df.isnull().sum()
missing_values
```

```python
# Tỷ lệ giá trị thiếu
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)
print("Tỷ lệ giá trị thiếu (%):")
print(missing_percentage)
```

```python
# Kiểm tra số dòng chứa giá trị null
num_null_rows = df.isnull().any(axis=1).sum()
print(f"Số dòng chứa giá trị null: {num_null_rows}")
```

```python
# Lọc các dòng có chứa giá trị null
rows_with_null = df[df.isnull().any(axis=1)]
print(rows_with_null)
```

# Xử lý dữ liệu

### Cột mssv

```python
# kiểm tra dữ liệu khuyết
missing_count = df['mssv'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mssv': {missing_count}")
```

```python
# Loại bỏ các cột bị khuyết
df.dropna(subset=['mssv'], inplace=True)
```

### cột mamh

```python
# Số lượng dữ liệu khuyết
missing_count = df['mamh'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'mamh': {missing_count}")
```

```python
# Các giá trị duy nhất và tần suất trong cột mamh
print(f"\nGiá trị duy nhất trong cột 'mamh':")
print(df['mamh'].value_counts())
print(df['mamh'].unique())
```

```python
# Bỏ khoảng trắng và viết thường trong các môn học
df['mamh'] = df['mamh'].str.replace(" ", "").str.lower().str.strip()
```

### cột sotc

```python
# Số lượng khuyết
missing_count = df['sotc'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'sotc': {missing_count}")
```

### cột namhoc

```python
# Số lượng khuyết
missing_count = df['namhoc'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'namhoc': {missing_count}")
```

### cột diemthi

```python
# Số lượng khuyết
missing_count = df['diem'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'diem': {missing_count}")
```

```python
df['diem'].fillna(df['diem'].mean(), inplace=True)
```

### cột trangthai

```python
# Số lượng khuyết
missing_count = df['diem'].isnull().sum()
print(f"Số lượng dữ liệu khuyết trong cột 'diem': {missing_count}")
```
